32)Implement the lasso model

```python
import numpy as np
from sklearn.linear_model import Lasso
X=np.array([[1,5,2],[2,6,4],[4,3,6],[7,6,2]])
Y=np.array([5,10,15,20])
model=Lasso(alpha=0.1)
model.fit(X,Y)
print("Coefficient:",model.coef_)
```

33)Implement the similarities and distances for any data set


```python
import numpy as np
A=np.array([[2,3],[3,4],[4,2]])
B=np.array([[8,7],[9,6],[10,8]])
P=np.array([5,5])
cA=A.mean(axis=0)
cB=B.mean(axis=0)
dA=np.linalg.norm(P-cA)
dB=np.linalg.norm(P-cB)
if dA<dB:
  print("Point belongs to cluster A")
else:
  print("Point belongs to cluster B")
```

34)Write a code to implement K-means clustering (unsupervised learning)algorithm

```python
import numpy as np
from scipy.spatial import distance
from scipy.stats import pearsonr
p1=np.array([2,3])
p2=np.array([5,7])
euc=distance.euclidean(p1,p2)
mh=distance.cityblock(p1,p2)
cb=distance.chebyshev(p1,p2)
hm=distance.hamming(p1,p2)
cd=distance.cosine(p1,p2)
cs=1-cd
set1,set2=set(p1),set(p2)
js=len(set1.intersection(set2))/len(set1.union(set2))
r,_=pearsonr(p1,p2)
print("Euclidean Distance:",euc)
print("Manhattan Distance:",mh)
print("Chebyshev Distance:",cb)
print("Hamming Distance:",hm)
print("Cosine Similarity:",cs)
print("Pearson Correlation:",r)
print("Jaccard Similarity:",js)
```
